# Udemy Data Analysis - Solutions


### Question 1
Q1: Load the dataset `udemy_courses.csv` into a Pandas DataFrame named `df` and display the first 5 rows along with the summary of column data types and non-null values.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

# Load dataset
df = pd.read_csv('udemy_courses.csv')
print(df.head())
print(df.info())
```

### Question 2
Q2: Check for missing values in each column and display total null values per column. Also display summary statistics for numeric columns.


```python
print('Missing values per column:')
print(df.isnull().sum())
print('\nSummary Statistics:')
print(df.describe())
```

### Question 3
Q3: Identify duplicate rows based on `course_id`. Print the count of duplicates and drop duplicate records, keeping the first occurrence.


```python
duplicates_count = df.duplicated(subset=['course_id']).sum()
print(f'Number of duplicate course_ids: {duplicates_count}')
df = df.drop_duplicates(subset=['course_id']).reset_index(drop=True)
print(f'New dataset shape: {df.shape}')
```

### Question 4
Q4: Convert `published_timestamp` to a datetime data type and create two new columns: `year` and `month_name`.


```python
df['published_timestamp'] = pd.to_datetime(df['published_timestamp'])
df['year'] = df['published_timestamp'].dt.year
df['month_name'] = df['published_timestamp'].dt.strftime('%B')
print(df[['published_timestamp', 'year', 'month_name']].head())
```

### Question 5
Q5: Convert `is_paid` boolean values into text category ('Paid' vs 'Free') in a new column called `course_type` and print the value counts.


```python
df['course_type'] = df['is_paid'].map({True: 'Paid', False: 'Free'})
print(df['course_type'].value_counts())
```

### Question 6
Q6: Create a count plot displaying the total number of courses available in each subject using Seaborn.


```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='subject', palette='Set2')
plt.title('Total Number of Courses by Subject', fontsize=14)
plt.xlabel('Subject')
plt.ylabel('Course Count')
plt.xticks(rotation=15)
plt.show()
```

### Question 7
Q7: Plot the distribution of course prices using a Seaborn histogram with KDE overlay. Exclude free courses ($0).


```python
plt.figure(figsize=(10, 5))
paid_courses = df[df['price'] > 0]
sns.histplot(paid_courses['price'], kde=True, bins=20, color='teal')
plt.title('Distribution of Course Prices (Paid Courses)', fontsize=14)
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.show()
```

### Question 8
Q8: Plot a pie chart showing the proportion of courses across difficulty levels (`level`).


```python
plt.figure(figsize=(7, 7))
level_counts = df['level'].value_counts()
plt.pie(level_counts, labels=level_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Proportion of Courses by Difficulty Level', fontsize=14)
plt.show()
```

### Question 9
Q9: Plot a histogram of `content_duration` in hours. Filter out extreme outliers (duration > 50 hours) for better visualization.


```python
plt.figure(figsize=(10, 5))
sns.histplot(df[df['content_duration'] <= 50]['content_duration'], bins=30, kde=True, color='purple')
plt.title('Distribution of Course Duration (<= 50 Hours)', fontsize=14)
plt.xlabel('Duration (Hours)')
plt.ylabel('Count')
plt.show()
```

### Question 10
Q10: Visualize the proportion of Paid vs Free courses using a Seaborn countplot with custom colors.


```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='course_type', palette=['#4C72B0', '#55A868'])
plt.title('Proportion of Free vs Paid Courses', fontsize=14)
plt.xlabel('Course Type')
plt.ylabel('Count')
plt.show()
```

### Question 11
Q11: Plot a boxplot of `num_subscribers` to identify potential outliers across the overall dataset.


```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['num_subscribers'], color='coral')
plt.title('Boxplot of Number of Subscribers', fontsize=14)
plt.xlabel('Subscribers')
plt.show()
```

### Question 12
Q12: Create a bar plot comparing total subscribers per subject.


```python
plt.figure(figsize=(10, 5))
subject_subscribers = df.groupby('subject')['num_subscribers'].sum().reset_index()
sns.barplot(data=subject_subscribers, x='subject', y='num_subscribers', palette='Blues_d')
plt.title('Total Subscribers by Subject', fontsize=14)
plt.xlabel('Subject')
plt.ylabel('Total Subscribers')
plt.xticks(rotation=15)
plt.show()
```

### Question 13
Q13: Plot a grouped bar chart showing the count of courses per subject divided by `course_type` (Free vs Paid).


```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='subject', hue='course_type', palette='viridis')
plt.title('Course Count per Subject by Type (Free vs Paid)', fontsize=14)
plt.xlabel('Subject')
plt.ylabel('Count')
plt.xticks(rotation=15)
plt.show()
```

### Question 14
Q14: Create a boxplot to compare course duration (`content_duration`) across different difficulty levels (`level`). Filter duration <= 40 hours.


```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[df['content_duration'] <= 40], x='level', y='content_duration', palette='Set3')
plt.title('Course Duration Distribution across Difficulty Levels', fontsize=14)
plt.xlabel('Difficulty Level')
plt.ylabel('Content Duration (Hours)')
plt.show()
```

### Question 15
Q15: Generate a scatter plot showing the relationship between `num_reviews` and `num_subscribers` using Matplotlib or Seaborn.


```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x='num_reviews', y='num_subscribers', alpha=0.6, hue='subject')
plt.title('Number of Reviews vs. Number of Subscribers', fontsize=14)
plt.xlabel('Number of Reviews')
plt.ylabel('Number of Subscribers')
plt.show()
```

### Question 16
Q16: Calculate and visualize average price by subject using horizontal bar chart.


```python
plt.figure(figsize=(9, 5))
avg_price = df.groupby('subject')['price'].mean().sort_values()
sns.barplot(x=avg_price.values, y=avg_price.index, palette='magma')
plt.title('Average Course Price by Subject ($)', fontsize=14)
plt.xlabel('Average Price ($)')
plt.ylabel('Subject')
plt.show()
```

### Question 17
Q17: Create a violin plot showing the distribution of prices across different course levels.


```python
plt.figure(figsize=(10, 6))
sns.violinplot(data=df[df['is_paid']], x='level', y='price', palette='coolwarm')
plt.title('Price Distribution across Difficulty Levels (Paid Courses)', fontsize=14)
plt.xlabel('Difficulty Level')
plt.ylabel('Price ($)')
plt.show()
```

### Question 18
Q18: Plot a scatter plot of `num_lectures` vs `content_duration` with a linear regression trend line.


```python
plt.figure(figsize=(9, 6))
sns.regplot(data=df, x='num_lectures', y='content_duration', scatter_kws={'alpha':0.3, 'color':'gray'}, line_kws={'color':'red'})
plt.title('Number of Lectures vs. Content Duration (Hours)', fontsize=14)
plt.xlabel('Number of Lectures')
plt.ylabel('Content Duration (Hours)')
plt.show()
```

### Question 19
Q19: Plot the line chart of total course publications year by year.


```python
yearly_courses = df.groupby('year').size().reset_index(name='course_count')
plt.figure(figsize=(9, 5))
sns.lineplot(data=yearly_courses, x='year', y='course_count', marker='o', linewidth=2.5, color='darkgreen')
plt.title('Total Course Publications Over Time (Yearly)', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Number of Courses Published')
plt.grid(True)
plt.show()
```

### Question 20
Q20: Plot total subscriber growth trends over the years broken down by subject.


```python
yearly_sub_subject = df.groupby(['year', 'subject'])['num_subscribers'].sum().reset_index()
plt.figure(figsize=(11, 6))
sns.lineplot(data=yearly_sub_subject, x='year', y='num_subscribers', hue='subject', marker='o')
plt.title('Subscriber Growth Trends by Subject (Yearly)', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Total Subscribers')
plt.legend(title='Subject')
plt.show()
```

### Question 21
Q21: Plot the monthly distribution of course publications using a bar plot to see which months have highest course releases.


```python
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
monthly_counts = df['month_name'].value_counts().reindex(month_order).reset_index()
monthly_counts.columns = ['month', 'count']

plt.figure(figsize=(11, 5))
sns.barplot(data=monthly_counts, x='month', y='count', palette='cividis')
plt.title('Course Publications by Month', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

### Question 22
Q22: Create a pivot table and heatmap showing the number of courses published per year for each subject.


```python
pivot_subject_year = df.pivot_table(index='subject', columns='year', values='course_id', aggfunc='count', fill_value=0)
plt.figure(figsize=(10, 5))
sns.heatmap(pivot_subject_year, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Heatmap of Course Releases by Subject and Year', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Subject')
plt.show()
```

### Question 23
Q23: Calculate and plot total revenue generated per year (Revenue estimation = `price` * `num_subscribers`).


```python
df['estimated_revenue'] = df['price'] * df['num_subscribers']
yearly_revenue = df.groupby('year')['estimated_revenue'].sum().reset_index()

plt.figure(figsize=(9, 5))
sns.barplot(data=yearly_revenue, x='year', y='estimated_revenue', palette='rocket')
plt.title('Estimated Total Revenue by Year ($)', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Estimated Revenue ($)')
plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

### Question 24
Q24: Compute and plot the Correlation Heatmap among numerical columns (`price`, `num_subscribers`, `num_reviews`, `num_lectures`, `content_duration`).


```python
num_cols = ['price', 'num_subscribers', 'num_reviews', 'num_lectures', 'content_duration']
corr_matrix = df[num_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Numerical Features', fontsize=14)
plt.show()
```

### Question 25
Q25: Find the top 10 most popular courses based on subscriber count and display them in a horizontal bar chart.


```python
top10_subs = df.nlargest(10, 'num_subscribers')

plt.figure(figsize=(10, 6))
sns.barplot(data=top10_subs, x='num_subscribers', y='course_title', palette='Blues_r')
plt.title('Top 10 Most Subscribed Courses on Udemy', fontsize=14)
plt.xlabel('Subscribers')
plt.ylabel('Course Title')
plt.show()
```

### Question 26
Q26: Create a new feature `engagement_rate` defined as (`num_reviews` / `num_subscribers`) * 100 for courses with subscribers > 100. Display the top 10 most engaged courses.


```python
df_active = df[df['num_subscribers'] >= 100].copy()
df_active['engagement_rate'] = (df_active['num_reviews'] / df_active['num_subscribers']) * 100
top10_engaged = df_active.nlargest(10, 'engagement_rate')

plt.figure(figsize=(10, 6))
sns.barplot(data=top10_engaged, x='engagement_rate', y='course_title', palette='Greens_r')
plt.title('Top 10 Courses by Engagement Rate (%)', fontsize=14)
plt.xlabel('Engagement Rate (%)')
plt.ylabel('Course Title')
plt.show()
```

### Question 27
Q27: Compare average engagement rate across subjects using a bar chart.


```python
avg_engagement = df_active.groupby('subject')['engagement_rate'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(data=avg_engagement, x='subject', y='engagement_rate', palette='Purples_d')
plt.title('Average Course Engagement Rate (%) by Subject', fontsize=14)
plt.xlabel('Subject')
plt.ylabel('Average Engagement Rate (%)')
plt.xticks(rotation=15)
plt.show()
```

### Question 28
Q28: Group price into 4 price ranges ('Free ($0)', 'Low ($20-$50)', 'Medium ($55-$100)', 'High ($105+)') and plot average subscriber counts per price range.


```python
def price_category(price):
    if price == 0:
        return 'Free ($0)'
    elif price <= 50:
        return 'Low ($20-$50)'
    elif price <= 100:
        return 'Medium ($55-$100)'
    else:
        return 'High ($105+)'

df['price_range'] = df['price'].apply(price_category)
range_order = ['Free ($0)', 'Low ($20-$50)', 'Medium ($55-$100)', 'High ($105+)']
price_sub = df.groupby('price_range')['num_subscribers'].mean().reindex(range_order).reset_index()

plt.figure(figsize=(9, 5))
sns.barplot(data=price_sub, x='price_range', y='num_subscribers', palette='Spectral')
plt.title('Average Subscribers by Price Range', fontsize=14)
plt.xlabel('Price Range')
plt.ylabel('Average Subscribers')
plt.show()
```

### Question 29
Q29: Determine the top 5 highest earning courses (by estimated revenue) and plot them in a horizontal bar chart.


```python
top5_revenue = df.nlargest(5, 'estimated_revenue')

plt.figure(figsize=(10, 5))
sns.barplot(data=top5_revenue, x='estimated_revenue', y='course_title', palette='YlOrRd_r')
plt.title('Top 5 Highest Revenue Generating Courses ($)', fontsize=14)
plt.xlabel('Estimated Revenue ($)')
plt.ylabel('Course Title')
plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

### Question 30
Q30: Create a multi-plot subplot grid (2x2) showing summary dashboards: 1) Subject counts, 2) Free vs Paid counts, 3) Yearly publications, 4) Top 5 subjects by revenue.


```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Subject counts
sns.countplot(data=df, x='subject', ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Course Count by Subject')
axes[0, 0].tick_params(axis='x', rotation=25)

# 2. Free vs Paid
sns.countplot(data=df, x='course_type', ax=axes[0, 1], palette='Set1')
axes[0, 1].set_title('Free vs Paid Courses')

# 3. Yearly trend
yearly_counts = df.groupby('year').size()
axes[1, 0].plot(yearly_counts.index, yearly_counts.values, marker='o', color='purple')
axes[1, 0].set_title('Yearly Course Releases')
axes[1, 0].set_xlabel('Year')
axes[1, 0].grid(True)

# 4. Revenue by subject
rev_subject = df.groupby('subject')['estimated_revenue'].sum().reset_index()
sns.barplot(data=rev_subject, x='subject', y='estimated_revenue', ax=axes[1, 1], palette='mako')
axes[1, 1].set_title('Total Revenue by Subject')
axes[1, 1].tick_params(axis='x', rotation=25)

plt.tight_layout()
plt.show()
```
