## Step 4 – Detect Missing Values

First calculate missing values with Pandas.

Then notice that some missing values are not actually stored as `NaN`. Values such as `NA`, `N/A`, `null`, empty strings and `Not Available` are also present.

### Student Task
Find:
- missing count per column
- missing percentage per column


```python
missing_count = df.isna().sum()
missing_percent = (df.isna().mean() * 100).round(2)

missing_report = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percent": missing_percent
}).sort_values("missing_percent", ascending=False)

missing_report
```

### Standardize Missing-Value Tokens

```python
missing_tokens = ["", "NA", "N/A", "null", "Not Available", "unknown"]

df = df.replace(missing_tokens, np.nan)

print("Total missing values:", df.isna().sum().sum())
df.isna().sum().sort_values(ascending=False)


```

## Step 5 – Decide How to Handle Missing Values

There is no single rule: the decision depends on the column.

For this dataset:
- `email` → keep missing if unavailable
- `phone` → keep missing if unavailable
- `gender`, `city`, `department` → fill with `"Unknown"`
- `age`, `marks`, `attendance` → use a sensible numeric strategy
- `annual_salary` → median is safer than mean when outliers exist
- `admission_date` → keep missing if it cannot be recovered



```python
# Convert numeric columns to numeric data type
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["marks"] = pd.to_numeric(df["marks"], errors="coerce")
df["attendance"] = pd.to_numeric(df["attendance"], errors="coerce")
df["annual_salary"] = pd.to_numeric(df["annual_salary"], errors="coerce")
```

```python
for col in ["gender", "city", "department", "course"]:
    df[col] = df[col].fillna("Unknown")

df["age"] = df["age"].fillna(df["age"].median())
df["marks"] = df["marks"].fillna(df["marks"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].median())
df["annual_salary"] = df["annual_salary"].fillna(df["annual_salary"].median())

df.isna().sum()
```

## Step 6 – Remove Extra Spaces and Standardize Text

A common real-world problem is that:

`Delhi`, ` Delhi`, `DELHI`, `delhi `

may represent the same category.

### Student Task
Clean the string columns using:
- `.str.strip()`
- `.str.lower()`

Then standardize known category variations.


```python
text_columns = ["student_name", "email", "gender", "city", "department", "course"]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

# Make selected categories lowercase for easier standardization
for col in ["gender", "city", "department", "course"]:
    df[col] = df[col].str.lower()

df[["student_name", "gender", "city", "department", "course"]].head(10)
```

```python
# Standardize category names
df["city"] = df["city"].replace({
    "bangalore": "bengaluru",
    "gurgaon": "gurugram"
})

df["department"] = df["department"].replace({
    "computer science": "cse",
    "information technology": "it",
    "electronics": "ece",
    "data science": "data science",
    "mechanical": "me"
})

df["course"] = df["course"].str.replace(" ", "", regex=False).str.lower()
df["course"] = df["course"].replace({
    "b.tech": "btech",
    "btech": "btech"
})

df[["city", "department", "course"]].value_counts().head(20)
```

## Step 7 – Find and Remove Duplicate Records

### Student Task
1. Find the number of completely duplicated rows.
2. Remove them.
3. Check the shape again.

Then discuss: **Is every repeated student name necessarily a duplicate?** No. We need a business key such as `student_id`.


```python
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)


```

```python
df.drop_duplicates(inplace=True)

print("New shape:", df.shape)
print("Duplicates remaining:", df.duplicated().sum())
```

## Step 8 – Clean Numeric Columns

The columns `age`, `marks`, `annual_salary`, and `attendance` may contain invalid values.

### Business rules
- Age should be between **17 and 35**
- Marks should be between **0 and 100**
- Attendance should be between **0 and 100**
- Salary should be greater than **0**

### Student Task
Identify invalid values first. Do not blindly delete rows.


```python
numeric_cols = ["age", "marks", "annual_salary", "attendance"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_cols].describe())
```

```python
# Trainer solution: convert impossible values to NaN
df.loc[~df["age"].between(17, 35), "age"] = np.nan
df.loc[~df["marks"].between(0, 100), "marks"] = np.nan
df.loc[df["annual_salary"] <= 0, "annual_salary"] = np.nan
df.loc[~df["attendance"].between(0, 100), "attendance"] = np.nan

# Refill values after invalid records were converted to NaN
df["age"] = df["age"].fillna(df["age"].median())
df["marks"] = df["marks"].fillna(df["marks"].median())
df["annual_salary"] = df["annual_salary"].fillna(df["annual_salary"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].median())
```

## Step 9 – Clean Email and Phone Numbers

### Email
A basic validation rule for this class:
- must contain `@`
- must contain `.`
- no spaces

### Phone
For this exercise, Indian student phone numbers should contain exactly **10 digits** and start with `6`, `7`, `8`, or `9`.

Do not build a perfect production validator yet. The goal is to understand cleaning logic.


```python
# Email cleaning
df["email"] = df["email"].str.strip().str.lower()

invalid_email = ~df["email"].str.match(r"^[^\s@]+@[^\s@]+\.[^\s@]+$", na=False)
print("Invalid emails:", invalid_email.sum())

df.loc[invalid_email, "email"] = np.nan

# Phone cleaning: remove spaces, hyphens and other non-digit characters
df["phone"] = df["phone"].astype("string").str.replace(r"\D", "", regex=True)

invalid_phone = ~df["phone"].str.match(r"^[6-9]\d{9}$", na=False)
print("Invalid phones:", invalid_phone.sum())

df.loc[invalid_phone, "phone"] = np.nan
```
